```python
import numpy as np

# AND Gate Dataset
X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]])

y = np.array([0,0,0,1])

# Initial weights and bias
w = np.array([0.5, 0.5])
b = -0.7

# Sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

print("Predictions:")

for i in range(len(X)):
    z = np.dot(X[i], w) + b
    output = sigmoid(z)
    prediction = 1 if output >= 0.5 else 0

    print("Input:", X[i],
          "Output:", round(output,4),
          "Prediction:", prediction)
```
