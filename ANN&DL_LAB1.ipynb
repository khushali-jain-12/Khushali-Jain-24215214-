```python
import numpy as np

def step(x):
    if x >= 0:
        return 1
    else:
        return 0


class Perceptron:
 # Initialize weights, bias, and learning rate.
    def __init__(self):
        self.weights = np.array([0.0, 0.0])
        self.bias = 0.0
        self.learning_rate = 0.1

    def predict(self, inputs):
        total = np.dot(inputs, self.weights) + self.bias
        return step(total)

    def train(self, X, y, epochs):
        for _ in range(epochs):
            for inputs, target in zip(X, y):

                prediction = self.predict(inputs)
                error = target - prediction


               # Update weights and bias
                self.weights = self.weights + self.learning_rate * error * inputs
                self.bias = self.bias + self.learning_rate * error


# AND gate dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 0, 0, 1])

# Create object and train
p = Perceptron()
p.train(X, y, 10)

# Test the model
print("Predictions:")
for data in X:
    print(data, "->", p.predict(data))
```
