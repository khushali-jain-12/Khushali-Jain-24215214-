```python
import torch
import torch.nn as nn
import torch.optim as optim

# AND Gate Dataset
X = torch.tensor([[0,0],
                  [0,1],
                  [1,0],
                  [1,1]], dtype=torch.float32)

y = torch.tensor([[0],
                  [0],
                  [0],
                  [1]], dtype=torch.float32)

# Create Neural Network
class NeuralNetwork(nn.Module):
    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(2, 4)   # Input → Hidden
        self.relu = nn.ReLU()           # Activation
        self.layer2 = nn.Linear(4, 1)   # Hidden → Output
        self.sigmoid = nn.Sigmoid()     # Output Activation

    def forward(self, x):
        x = self.layer1(x)
        x = self.relu(x)
        x = self.layer2(x)
        x = self.sigmoid(x)
        return x

# Create model
model = NeuralNetwork()

# Loss function and Optimizer
loss_function = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

# Train the model
for epoch in range(100):
    output = model(X)
    loss = loss_function(output, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

# Predict
print("Predictions:")

with torch.no_grad():
    prediction = model(X)

    for i in range(len(X)):
        print(X[i].tolist(), "->", int(prediction[i] >= 0.5))
```

```python
# LAB 3: Feedforward Neural Network using TensorFlow

import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# AND Gate Dataset
X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]])

y = np.array([[0],
              [0],
              [0],
              [1]])

# Create Neural Network
model = Sequential()

model.add(Dense(4, input_dim=2, activation='relu'))
model.add(Dense(1, activation='sigmoid'))

# Compile Model
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Train Model
model.fit(X, y, epochs=100, verbose=0)

# Predict
print("Predictions:")

prediction = model.predict(X)

for i in range(len(X)):
    print(X[i], "->", int(prediction[i] >= 0.5))
```
